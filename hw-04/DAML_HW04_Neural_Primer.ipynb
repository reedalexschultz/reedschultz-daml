## 1. Read a shape before computing

An array's shape records the number of entries along each axis. The names assigned to those axes come from the problem, not from NumPy.

For the array below, rows are observations and columns are measurements. Predict the shape of a single row, a one-row slice, and a single column. Why might keeping a one-row array matter when combining operations?

```python
observations = np.array([[2., 5., 1.], [4., 3., 7.]])
print("all observations:", observations.shape)
print("one row:", observations[0].shape)
print("one-row slice:", observations[0:1].shape)
print("one column:", observations[:, 1].shape)
print("transposed:", observations.T.shape)
```

## 2. Broadcasting as a shape rule

Broadcasting lets compatible arrays participate in elementwise operations. This example converts measurements between units; it does not build a neural layer.

Before running it, predict whether the same scale is applied to a row or a column. Change one scale factor and check exactly which entries change.

```python
measurements = np.array([[1., 2., 3.], [4., 5., 6.]])
unit_scales = np.array([10., 100., 1000.])
converted = measurements * unit_scales
print("input shape:", measurements.shape)
print("scale shape:", unit_scales.shape)
print(converted)
```

## 3. Select a patch and label its coordinates

Slicing is useful for inspecting a local region of an image. Python slice endpoints are exclusive.

Predict which numbers appear in the highlighted patch. Move the patch one column right, then change its height. Describe which coordinates you changed and why. This exercise selects a patch; deciding how a graded kernel transforms it is your homework task.

```python
image = np.arange(30).reshape(5, 6)
row, col = 1, 2
height, width = 2, 3
patch = image[row:row + height, col:col + width]
print("image:", image)
print("selected patch:", patch)
fig, ax = plt.subplots(figsize=(5, 3))
ax.imshow(image, cmap="gray")
ax.add_patch(plt.Rectangle((col - .5, row - .5), width, height,
                           fill=False, edgecolor="orange", linewidth=3))
ax.set_title("Selected region")
plt.show()
```

## 4. Axes and retained dimensions

Many array methods accept `axis`. Learn what it means with a counting task unrelated to attention normalization.

Predict each output's shape. Then explain why the column counts have a different shape from the row counts. `keepdims=True` retains a length-one axis where an axis was reduced.

```python
flags = np.array([[True, False, True], [False, False, True]])
print("flags shape:", flags.shape)
print("counts by row:", np.count_nonzero(flags, axis=1))
print("counts by column:", np.count_nonzero(flags, axis=0))
print("row counts with retained axis:",
      np.count_nonzero(flags, axis=1, keepdims=True).shape)
```

## 5. Floating-point precision is finite

Two distinct mathematical values can become indistinguishable when represented on a computer. Predict the result below, then try smaller and larger magnitudes.

When you write a numerical routine, consider both small hand-checkable inputs and unusually large inputs. This experiment diagnoses representation limits; it does not provide a normalization algorithm.

```python
large = np.float64(1e16)
print("spacing near this value:", np.spacing(large))
print("represented increment:", (large + 1.0) - large)
print("increment at a smaller scale:", (np.float64(1e6) + 1.0) - 1e6)
```

## 6. Turn a claim into a check

Checks should express a property you expect, not merely repeat your implementation. Below is an ungraded array-copy exercise.

Before running, predict whether editing `changed` will alter `original`. Then design one additional check for a different input.

```python
original = np.array([2., 4., 6.])
changed = original.copy()
changed[0] = 99.
assert original.shape == changed.shape
assert np.array_equal(original, np.array([2., 4., 6.]))
assert changed[0] == 99.
print("Copy and shape checks passed.")
```
